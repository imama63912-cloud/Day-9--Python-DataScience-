```python
import numpy as np
import pandas as pd

file_path = r"C:\Users\SK\Desktop\New folder\student performance.csv"

# Dataset
df = pd.read_csv(file_path)
df.head()

```

```python
# Check structure
print(df.head())
print("\nShape of DataFrame:", df.shape)

# Extract numeric columns for arrays
math = df["math score"].to_numpy()
reading = df["reading score"].to_numpy()
writing = df["writing score"].to_numpy()

# One-dimensional arrays
print("\n=== 1D Arrays ===")
print("Math scores   → shape:", math.shape, "| ndim:", math.ndim, "| size:", math.size, "| dtype:", math.dtype)
print("Reading scores → shape:", reading.shape)
print("Writing scores → shape:", writing.shape)

# Two-dimensional array
scores_2d = df[["math score", "reading score", "writing score"]].to_numpy()
print("\n=== 2D Array ===")
print("scores_2d → shape:", scores_2d.shape, "| ndim:", scores_2d.ndim, "| size:", scores_2d.size, "| dtype:", scores_2d.dtype)
print(scores_2d[:5])   # first 5 rows
```

```python
# Extract numeric columns for arrays
math = df["math score"].to_numpy()
reading = df["reading score"].to_numpy()
writing = df["writing score"].to_numpy()

# One-dimensional arrays
print("1D Math scores array:")
print(math)
print("Shape:", math.shape, "| ndim:", math.ndim, "| size:", math.size, "| dtype:", math.dtype)

print("\n1D Reading scores array:")
print(reading[:10], "...")  # first 10 for brevity
print("Shape:", reading.shape, "| ndim:", reading.ndim, "| size:", reading.size, "| dtype:", reading.dtype)

# Two-dimensional array (all three scores side-by-side)
scores_2d = df[["math score", "reading score", "writing score"]].to_numpy()
print("\n2D Scores array (math | reading | writing):")
print(scores_2d[:5])  # first 5 rows
print("Shape:", scores_2d.shape, "| ndim:", scores_2d.ndim, "| size:", scores_2d.size, "| dtype:", scores_2d.dtype)
```

```python
print("=== Array Attributes ===")
print(f"math.shape   = {math.shape}")      # (1000,)
print(f"math.ndim    = {math.ndim}")       # 1
print(f"math.size    = {math.size}")       # 1000
print(f"math.dtype   = {math.dtype}")      # int64

print(f"\nscores_2d.shape = {scores_2d.shape}")  # (1000, 3)
print(f"scores_2d.ndim  = {scores_2d.ndim}")     # 2
print(f"scores_2d.size  = {scores_2d.size}")     # 3000
print(f"scores_2d.dtype = {scores_2d.dtype}")    # int64
```

```python
# Element-wise arithmetic
total = math + reading + writing
average = total / 3

print("First 5 total scores:", total[:5])
print("First 5 averages   :", np.round(average[:5], 2))

# Statistical operations
print("\nMath score stats:")
print(f"  Mean   : {np.mean(math):.2f}")
print(f"  Median : {np.median(math):.2f}")
print(f"  Std    : {np.std(math):.2f}")
print(f"  Min    : {np.min(math)}")
print(f"  Max    : {np.max(math)}")

# Boolean masking / filtering
high_math = math[math >= 80]
print(f"\nNumber of students with math ≥ 80: {len(high_math)}")
print("Their scores:", high_math[:10], "...")

# Comparison with pure Python lists (performance & syntax)
py_list = math.tolist()
# NumPy is vectorized – much faster and cleaner for large data
print("\nNumPy mean:", np.mean(math))
print("Python mean (manual):", sum(py_list)/len(py_list))
```
